```python
# NZ major cities
cities = ["Christchurch", "Wellington", "Auckland", "Dunedin", "Hamilton"]
pops_2024 = [404700, 215400, 1695200, 126300, 189700]
lats = [-43.5321, -41.2924, -36.8485, -45.8742, -37.7870]
lons = [172.6362, 174.7787, 174.7633, 170.5036, 175.2793]
founded_years = [1856, 1840, 1840, 1848, 1864]

NZ_TOTAL_POP = 5_347_000
```

```python
for i in range(len(cities)):
    intro = f"{cities[i]} was founded in {founded_years[i]}, "
    intro += f"population {pops_2024[i]:,}, "
    intro += f"coordinates ({lats[i]:.4f}, {lons[i]:.4f})"
    print(intro)
```

```python
south_island = [city for i,city in enumerate(cities) if lats[i] < -41]
print(f"South Island cities: {south_island}")

upper_names = [city.upper() for city in cities]
print(upper_names)

pct_of_nz = [f"{p / NZ_TOTAL_POP * 100:.1f}%" for p in pops_2024]
print(pct_of_nz)
```

```python
regions = ["Canterbury", "Wellington", "Auckland", "Otago", "Waikato"]

# Use a list comprehension to filter region names that start with "A" or "O".
# Expected: ['Auckland', 'Otago']
a_or_o_regiond = [
    region for region in regions if region.startswith("A") or region.startswith("O")
]
print(a_or_o_regiond)
```

```python
# Create a dictionary for your most interested NZ city.
# Include at least: name, lat, lon, why_interested
# Print a one-sentence intro using f-string.
most_interested_city = "Christchurch"
city_index = cities.index(most_interested_city)
most_interested_city_data = {
    "name": most_interested_city,
    "lat": lats[city_index],
    "lon": lons[city_index],
    "why_interested": "It is a coastal and beautiful city.",
}
intro = f"My most interested city is {most_interested_city_data['name']}, "
intro += f"its coordinates are ({most_interested_city_data['lat']:.4f}, {most_interested_city_data['lon']:.4f}), "
intro += f"{most_interested_city_data['why_interested']}"
print(intro)
```

```python
# Use zip() to combine `cities` and `pops_2024` into a list of (city, population) tuples.
# Then sort the list by population in descending order.
# Expected first item: ('Auckland', 1695200)
city_populations = sorted(
    zip(cities, pops_2024), key=lambda x: x[1], reverse=True
)  # default ascending order
print(city_populations)
```
